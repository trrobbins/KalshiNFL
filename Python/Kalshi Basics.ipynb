```python
import os
import pandas as pd

def GetTradeFiles(trade_path: str = r"E:\PredMktData\TradeExports") -> pd.DataFrame:
    """
    Retrieve a list of all trade CSV files from a specified directory
    and return them as a pandas DataFrame.

    The function identifies files ending with '-Trades.csv', constructs
    their full file paths, and derives a corresponding ticker name by
    removing the '-Trades.csv' suffix.

    Parameters
    ----------
    trade_path : str, optional
        The directory containing trade export files.
        Defaults to 'E:\\PredMktData\\TradeExports'.

    Returns
    -------
    pandas.DataFrame
        A DataFrame with the following columns:
        - 'Filename' : the CSV filename (e.g., 'KXNFLGAME-25SEP04DALPHI-PHI-Trades.csv')
        - 'FullPath' : the full path to the file
        - 'Ticker'   : the ticker extracted from the filename
                       (e.g., 'KXNFLGAME-25SEP04DALPHI-PHI')

    Raises
    ------
    FileNotFoundError
        If the specified trade_path directory does not exist.

    Examples
    --------
    >>> df_trade = GetTradeFiles()
    >>> print(df_trade.head())

    >>> df_custom = GetTradeFiles(trade_path=r"D:\\MyProject\\Data\\Trades")
    >>> print(len(df_custom))
    """

    # ------------------------------------------------------------------
    # 1. Validate directory existence
    # ------------------------------------------------------------------
    if not os.path.exists(trade_path):
        raise FileNotFoundError(f"The directory does not exist: {trade_path}")

    # ------------------------------------------------------------------
    # 2. Collect all filenames ending with '-Trades.csv'
    # ------------------------------------------------------------------
    trade_files = [
        f for f in os.listdir(trade_path)
        if f.endswith("-Trades.csv")
    ]

    # ------------------------------------------------------------------
    # 3. Construct DataFrame with filenames and paths
    # ------------------------------------------------------------------
    df_trade = pd.DataFrame({
        "Filename": trade_files,
        "FullPath": [os.path.join(trade_path, f) for f in trade_files]
    })

    # ------------------------------------------------------------------
    # 4. Derive ticker name from filename
    # ------------------------------------------------------------------
    df_trade["Ticker"] = df_trade["Filename"].str.replace("-Trades.csv", "", regex=False)

    # ------------------------------------------------------------------
    # 5. Provide summary and return
    # ------------------------------------------------------------------
    print(f"Found {len(df_trade)} trade files in {trade_path}")
    return df_trade


```

```python
# This will create 1 minute aggregates for all games
import pandas as pd
import os

trade_path = r"E:\PredMktData\TradeExports"
ohlc_path = r"E:\PredMktData\OHLCExports"
df_trade = GetTradeFiles()


for ticker, fn in zip(df_trade['Ticker'], df_trade['Filename']):
    # --- Skip files that don't match your trade naming pattern ---
    if not str(fn).endswith("-Trades.csv"):
        print(f"Skipping {fn} (not a trade file)")
        continue

    print(f"\nProcessing {fn} for {ticker}...")

    file_path = os.path.join(trade_path, fn)

    # --- Skip if the file does not actually exist ---
    if not os.path.exists(file_path):
        print(f"  File not found: {file_path}")
        continue

    # --- Load CSV into DataFrame ---
    try:
        trades_df = pd.read_csv(file_path, parse_dates=["created_time"])
    except Exception as e:
        print(f"  Error reading {fn}: {e}")
        continue

    # --- Confirm basic info ---
    print(f"  Loaded {len(trades_df)} rows")

    # --- Generate 1-hour OHLC data ---
    try:
        ohlc_1h = ohlc_from_trades(
            trades_df=trades_df,
            freq="1H",
            save_file=True,
            save_path=ohlc_path
        )
        print(f"  Saved {len(ohlc_1h)} hourly bars to {ohlc_path}")
    except Exception as e:
        print(f"  Error processing {fn}: {e}")
        continue
```

```python
import requests

BASE_URL = "https://api.elections.kalshi.com/trade-api/v2"

def find_sport_series(sport: str, scope: str | None = None):
    params = {"category": "Sports"}
    if sport and scope:
        # The API expects a single comma-separated string for tags
        params["tags"] = f"{sport},{scope}"
    elif sport:
        params["tags"] = sport

    r = requests.get(f"{BASE_URL}/series", params=params, timeout=30)
    r.raise_for_status()
    data = r.json() or {}
    ser = data.get("series") or []   # <-- key change: coalesce None to []
    return [(s.get("ticker"), s.get("title"), s.get("tags")) for s in ser]



def get_sports_filters():
    r = requests.get(f"{BASE_URL}/search/filters_by_sport", timeout=30)
    r.raise_for_status()
    return r.json().get("filters_by_sports", {})

def list_series(tags: str, category: str = "Sports", limit: int = 200):
    """List series for given comma-separated tags; handles null and pagination."""
    params = {"category": category, "tags": tags, "limit": str(limit)}
    out, cursor = [], None
    while True:
        if cursor:
            params["cursor"] = cursor
        r = requests.get(f"{BASE_URL}/series", params=params, timeout=30)
        r.raise_for_status()
        data = r.json() or {}
        series = data.get("series") or []  # coalesce None -> []
        out.extend(series)
        cursor = data.get("cursor")
        if not cursor:
            break
    return out

def find_series_for(competition: str, scope: str = "Games"):
    """
    Try tag combinations in a sensible order using the official strings:
      1) 'Football,College Football,Games'  (Sport,Competition,Scope)
      2) 'College Football,Games'           (Competition,Scope)
      3) 'College Football'                 (Competition only)
    Return a list of (ticker, title, tags).
    """
    attempts = [
        f"Football,{competition},{scope}",
        f"{competition},{scope}",
        f"{competition}",
    ]
    results = []
    for tags in attempts:
        series = list_series(tags)
        if series:
            results.extend(series)
    # Deduplicate by ticker
    seen, deduped = set(), []
    for s in results:
        t = s.get("ticker")
        if t and t not in seen:
            seen.add(t)
            deduped.append((t, s.get("title"), s.get("tags")))
    return deduped

```

```python
def get_historical_markets_by_series(series_ticker: str, limit=500):
    params = {
        "series_ticker": series_ticker,
        "limit": str(limit),
    }
    markets, cursor = [], None
    while True:
        if cursor:
            params["cursor"] = cursor
        r = requests.get(f"{BASE_URL}/historical/markets", params=params, timeout=30)
        r.raise_for_status()
        data = r.json()
        markets.extend(data.get("markets", []))
        cursor = data.get("cursor")
        if not cursor:
            break
    return pd.json_normalize(markets)

def get_nfl_historical_games_df(limit=500):
    return get_historical_markets_by_series(series_ticker="KXNFLGAME", limit=limit)
```
